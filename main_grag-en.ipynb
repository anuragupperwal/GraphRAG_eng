```python
# src/eval/evaluate_existing_results.py
import os
import re
import faiss
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

# --------------------------------------------------------------------
# CONFIGURATION (adjust paths as per your Kaggle environment)
# --------------------------------------------------------------------
GRAPH_DIR = "/kaggle/working/GraphRAG_eng/data/knowledge_graph"
OUT_DIR = "/kaggle/working/GraphRAG_eng/data/output"
EMB_MODEL = "nomic-ai/nomic-embed-text-v1.5"
RESULT_PATH = os.path.join(OUT_DIR, "evaluation_report_refreshed.csv")

# --------------------------------------------------------------------
# UTILITIES
# --------------------------------------------------------------------

def load_index_and_meta():
    """Load FAISS index and metadata."""
    faiss_path = os.path.join(GRAPH_DIR, "community_faiss.index")
    meta_path = os.path.join(GRAPH_DIR, "community_meta.pkl")

    assert os.path.exists(faiss_path), f"Missing {faiss_path}"
    assert os.path.exists(meta_path), f"Missing {meta_path}"

    index = faiss.read_index(faiss_path)
    meta = pd.read_pickle(meta_path)
    return index, meta


def recall_at_k(index, embeddings, k=5):
    """Self-retrieval Recall@K."""
    D, I = index.search(embeddings, k)
    correct = sum(i in I[i] for i in range(len(I)))
    return round(correct / len(I), 4)


def precision_at_k(index, embeddings, k=5, threshold=0.7):
    """Cosine-similarity-based approximate precision."""
    D, _ = index.search(embeddings, k)
    precision_scores = [np.sum(D[i] > threshold) / k for i in range(len(D))]
    return round(np.mean(precision_scores), 4)


def semantic_faithfulness(answer, contexts, model):
    """Embedding-based faithfulness proxy."""
    if not answer.strip():
        return 0.0
    ans_emb = model.encode([answer], normalize_embeddings=True)
    ctx_embs = model.encode(contexts, normalize_embeddings=True)
    sims = cosine_similarity(ans_emb, ctx_embs)
    return round(float(np.mean(sims)), 4)


def context_coverage(answer, contexts):
    """Lexical overlap between answer and retrieved context."""
    answer_words = set(re.findall(r"\w+", answer.lower()))
    context_words = set(re.findall(r"\w+", " ".join(contexts).lower()))
    overlap = len(answer_words & context_words)
    return round(overlap / len(answer_words), 4) if len(answer_words) > 0 else 0.0


# --------------------------------------------------------------------
# MAIN EVALUATION
# --------------------------------------------------------------------
if __name__ == "__main__":
    print("🔍 Loading FAISS index and metadata...")
    index, meta = load_index_and_meta()

    print("📈 Reconstructing or re-embedding summaries...")
    try:
        embs = np.vstack([index.reconstruct(i) for i in range(index.ntotal)])
    except Exception:
        print("⚠️ Could not reconstruct; recomputing embeddings.")
        model = SentenceTransformer(EMB_MODEL, trust_remote_code=True)
        embs = model.encode(meta["summary"].tolist(), normalize_embeddings=True).astype("float32")

    print("🧠 Loading embedding model...")
    model = SentenceTransformer(EMB_MODEL, trust_remote_code=True)

    ans_path = os.path.join(OUT_DIR, "answer.txt")
    assert os.path.exists(ans_path), f"Missing generated answer at {ans_path}"
    answer = open(ans_path).read().strip()
    contexts = meta["summary"].tolist()[:5]

    print("📊 Computing metrics...")
    recall_5 = recall_at_k(index, embs, k=5)
    precision_5 = precision_at_k(index, embs, k=5)
    faith = semantic_faithfulness(answer, contexts, model)
    coverage = context_coverage(answer, contexts)
    hallucination = round(1 - faith, 4)

    results = pd.DataFrame([{
        "Recall@5": recall_5,
        "Precision@5": precision_5,
        "Faithfulness": faith,
        "Context_Coverage": coverage,
        "Hallucination": hallucination
    }])
    os.makedirs(OUT_DIR, exist_ok=True)
    results.to_csv(RESULT_PATH, index=False)

    print("\n✅ Evaluation Summary:")
    print(results)
    print(f"\n💾 Saved report to {RESULT_PATH}")
```

```python
import os
import faiss
import pandas as pd
from sentence_transformers import SentenceTransformer
import google.generativeai as genai
from kaggle_secrets import UserSecretsClient

# ---------------- CONFIG ----------------
GRAPH_DIR = "data/knowledge_graph"
OUT_DIR = "data/output"
EMB_MODEL = "nomic-ai/nomic-embed-text-v1.5"
TOP_K = 5
MODEL_NAME = "gemini-2.5-flash"

QUERIES = [
    "What happened during Hurricane Katrina and which areas were most affected?",
    "Explain the conflict between Russia and Georgia in 2008.",
    "What did President Obama say about healthcare reform?",
    "Who won the 2008 US presidential election and why was it historic?",
    "How did the Mumbai terrorist attacks unfold in 2008?",
]

# ---------------- SETUP ----------------
os.makedirs(OUT_DIR, exist_ok=True)
user_secrets = UserSecretsClient()
genai.configure(api_key=user_secrets.get_secret("GOOGLE_API_KEY"))

index = faiss.read_index(os.path.join(GRAPH_DIR, "community_faiss.index"))
meta = pd.read_pickle(os.path.join(GRAPH_DIR, "community_meta.pkl"))
emb_model = SentenceTransformer(EMB_MODEL, trust_remote_code=True)
llm = genai.GenerativeModel(MODEL_NAME)

# ---------------- HELPERS ----------------
def embed_texts(texts):
    return emb_model.encode(texts, convert_to_numpy=True, normalize_embeddings=True).astype("float32")

def retrieve(query, top_k=5):
    q_emb = embed_texts([query])
    scores, idxs = index.search(q_emb, top_k)
    retrieved = meta.iloc[idxs[0]].copy()
    retrieved["score"] = scores[0]
    return retrieved

def generate_answer(query, context):
    prompt = (
        "You are an expert assistant within a retrieval-augmented system.\n"
        "Use *only* the provided context to answer the question faithfully and in detail.\n\n"
        f"Context:\n{context}\n\nQuestion: {query}\nAnswer:"
    )
    try:
        response = llm.generate_content(prompt)
        return response.text.strip()
    except Exception as e:
        print(f"[Gemini Error] {e}")
        return "(Error generating answer)"

# ---------------- RUN MULTI-QUERY TEST ----------------
results = []

for query in QUERIES:
    print(f"\n🔍 Query: {query}")
    retrieved = retrieve(query, top_k=TOP_K)

    print("\n📚 Top Retrieved Community Summaries:")
    for cid, row in retrieved.iterrows():
        print(f"--- [{row['community']}] (score={row['score']:.3f}) ---")
        print(row['summary'][:600], "\n")

    context = "\n\n".join(retrieved["summary"].tolist())
    answer = generate_answer(query, context)
    print("💬 Generated Answer:\n", answer[:1200], "\n")

    results.append({
        "query": query,
        "retrieved_ids": retrieved["community"].tolist(),
        "scores": [float(s) for s in retrieved["score"].tolist()],
        "answer": answer,
        "context": context[:1500],
    })

# ---------------- SAVE RESULTS ----------------
results_df = pd.DataFrame(results)
results_path = os.path.join(OUT_DIR, "query_results.csv")
results_df.to_csv(results_path, index=False, encoding="utf-8")

print(f"\n✅ Saved all query results to {results_path}")
display(results_df.head())
```
